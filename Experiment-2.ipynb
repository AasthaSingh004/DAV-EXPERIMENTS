```python
print("\nMissing Values:\n")
print(df.isnull().sum())
```

```python
print("\nRows with missing values:\n")
df[df.isnull().any(axis=1)]
```

```python
print("\nNumber of Duplicate Records:\n")
print(df.duplicated().sum())
```

```python
print("\nDuplicate Records:\n")
df[df.duplicated()]
```

```python
df=df.drop_duplicates()
print("Shape after removing duplicates:",df.shape)
```

```python
df["Attendance"]=df["Attendance"].fillna(df["Attendance"].mean())
df["Internal_Marks"]=df["Internal_Marks"].fillna(df["Internal_Marks"].mean())
print("\nMissing Values after filling:\n")
print(df.isnull().sum())
```

```python
print("\nAverage Exam Score by Department:\n")
df.groupby("Department")["Exam_Score"].mean()
```

```python
print("\nAverage Exam Score by Gender:\n")
df.groupby("Gender")["Exam_Score"].mean()
```

```python
print("\nPass and Fail count:\n")
df["Result"].value_counts()
```
