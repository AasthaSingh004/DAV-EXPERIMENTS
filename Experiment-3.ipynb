```python
import pandas as pd
df=pd.read_csv("student_performance.csv", index_col="S_No")
df=df.drop_duplicates()
df=df.fillna(df.mean(numeric_only=True))
df
```

```python
import matplotlib.pyplot as plt
plt.hist(df["Exam_Score"], bins=5)
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Score")
plt.show()
```

```python
plt.hist(df["Study_Hours"], bins=5)
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")
plt.title("Distribution of Study Hours")
plt.show()
```

```python
plt.boxplot(df["Exam_Score"])
plt.xlabel("Exam Score")
plt.ylabel("Marks")
plt.title("Box Plot of Exam Score")
plt.show()
```

```python
plt.boxplot([
    df["Study_Hours"],
    df["Attendance"],
    df["Assignment_Marks"],
    df["Internal_Marks"],
    df["Exam_Score"]
])
plt.xticks(
    [1, 2, 3, 4, 5],
    ["Study Hours", "Attendance", "Assignment", "Internal", "Exam"]
)
plt.ylabel("Values")
plt.title("Box Plot of Student Performance")
plt.show()
```

```python
plt.scatter(df["Study_Hours"], df["Exam_Score"])
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
```

```python
plt.scatter(df["Attendance"], df["Exam_Score"])
plt.xlabel("Attendance")
plt.ylabel("Exam Score")
plt.title("Attendance vs Exam Score")
plt.show()
```

```python
import seaborn as sns
sns.pairplot(
    df[
        [
            "Study_Hours",
            "Attendance",
            "Assignment_Marks",
            "Internal_Marks",
            "Exam_Score"
        ]
    ]
)
plt.show()
```

```python
sns.heatmap(
    df[
        [
            "Study_Hours",
            "Attendance",
            "Assignment_Marks",
            "Internal_Marks",
            "Exam_Score"
        ]
    ].corr(),
    annot=True
)
plt.title("Correlation Heatmap")
plt.show()
```
