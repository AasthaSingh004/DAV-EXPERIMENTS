```python
import numpy as np
import pandas as pd
df=pd.read_csv("student_performance.csv", index_col="S_No")
df=df.drop_duplicates()
arr=np.array(df["Exam_Score"].head(6))
print("\n Numpy array")
print("Array: ", arr)
```

```python
print("First element:",arr[0])
print("Second element:",arr[1])
print("Third element:",arr[2])
print("Fourth element:",arr[3])
print("Fifth element:",arr[4])
print("Last element:",arr[-1])
```

```python
print("Array Slicing\n")
print("first 3 elements:",arr[:3])
print("elements from index 3:",arr[3:])
print("last 2 elements:", arr[-2:])
```

```python
print("Basic array operations\n")
print("Add 10:", arr+10)
print("Subtract 5:", arr-5)
print("Multiply 2:", arr*2)
print("Divide 10:", arr/10)
```

```python
print("Mathematical operations\n")
print("Sum:",np.sum(arr))
print("Mean:",np.mean(arr))
print("Maximum:",np.max(arr))
print("Minimum:",np.min(arr))
```

```python
new_arr=arr.reshape(3, 2)
print("\nOriginal array:\n")
print(arr)
print("\nReshaped array:\n")
print(new_arr)
```

```python
print("\nPandas DataFrame\n")
print(df)
```

```python
print("Name Column:")
print(df["Name"])
print("Exam_Score Column:")
print(df["Exam_Score"])
```

```python
print("Name and Exam_Score Columns:")
print(df[["Name", "Exam_Score"]])
```

```python
print("Average Exam Score:",df["Exam_Score"].mean())
print("Highest Exam Score:",df["Exam_Score"].max())
print("Lowest Exam Score:",df["Exam_Score"].min())
```

```python
print("\nRow Indexing\n")
print("\nFirst Row:\n")
print(df.iloc[0])
print("\nSecond Row:\n")
print(df.iloc[1])
print("\nThird Row:\n")
print(df.iloc[2])
```

```python
print("\nSpecific Values\n")
print("Name in first row:", df.iloc[0,0])
print("Department in first row:", df.iloc[0,3])
print("Exam score of Amrutha:", df.iloc[1,8])
print("Attendance of Shivam:", df.iloc[2,4])
```

```python
print("\nRow Slicing\n")
print("First 5 rows:\n")
print(df.iloc[0:5])
print("\nRows 5 to 9, first 5 columns:\n")
print(df.iloc[5:10, 0:5])
```

```python
print("\nData Filtering\n")
result=df[df["Exam_Score"]>80]
print("Students with exam score greater than 80:\n")
print(result[["Name","Department","Exam_Score"]])
```

```python
print("\nFiltering with two conditions\n")
result=df[(df["Department"]=="CSE") & (df["Attendance"]>85)]
print(result[["Name","Department","Attendance"]])
```
